# Mapping migration

### An introduction to vector data operations

The Veery thrush (*Catharus fuscescens*) migrates each year between nesting sites in North America and South America. Veeries are small birds, and encountering a hurricane during their migration would be disasterous. However, Ornithologist Christopher Heckscher has found in tracking their migration that their breeding patterns and migration timing can predicting the severity of hurricane season as accurately as meterological models [(Heckscher 2018)](https://doi.org/10.1038/s41598-018-28302-3)!

### Reflect and Respond #1: Reflect on what you know about migration. You could consider:
1. What are some reasons that animals might migrate?
2. How might climate change affect animal migrations?
3. Do you notice any animal migrations where you live?

**Your answer here:**

1. Animals may migrate for a plethora of reasons depending on their habitat ranges, diet, breeding patterns, and external stresses from humans and climate change. For example, certain whale species like humpbacks and gray whales migrate between cold water polar regions where they feed and warm water tropical regions where they breed and reproduce. Alternatively, some species like monarch butterflies migrate seasonally to avoid unfavorable climates and access better food resources, with these butterflies migrating south in the winter to avoid freezing conditions that they and cannot survive or find nectar producing plants in.

2. Climate change already has, and will almost certainly continue to shift the migration patterns of many species. The primary cause of this is changing air and ocean temperatures, which result in different timings of species migration and different locations of migrations. Polar bears, for example, have begun migrating much further south and into populated areas as Arctic temperatures have increased, reducing the amount of sea ice polar bears have as their natural habitat, and thus reducing their ability to find food sources and literally pulling their land out from under their feet. 

3. In Colorado, I've noticed many bird migrations seasonally like humming birds and owls. I also spend a lot of time in Rocky Mountain National Park for field work, and I've gotten to see the migration of elk into higher elevations as temperatures become warmer to access the tundra shrubs for food. Back home in San Diego, I always loved when monarch butterflies would migrate northwards from Mexico in the spring, and great white shark sightings are always the highest in the late summer, which is both cool to witness and absolutely terrifying when I go surfing.
